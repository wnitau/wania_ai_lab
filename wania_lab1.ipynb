```python
destination = input("Where do you want to study?")
budget = int(input("What is your budget? "))
study_plans = {
    "canada": 800000,
    "hungary": 450000,
    "malaysia": 400000
}
print("\nAgent processig...")
print("please write in lower case")

if destination in study_plans:
    cost = study_plans[destination]

    if budget >= cost:
        print("Agent: Yes! You can study in", destination)
    else:
        print("Agent: This destination is over your budget.")
        print("Agent: Choose a cheaper destination.")
else:
    print("Agent: No suitable study plan found.")
```
